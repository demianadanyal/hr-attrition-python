```python
df = df.drop(columns=['Over18', 'Employee Count', 'Standard Hours', '-2', '0'])
df.shape
```

```python
#Q1
df.Attrition.value_counts()
```

```python
df.Attrition.value_counts(dropna=False, normalize=True)
```

```python
sns.catplot( data = df, x='Attrition', kind='count' )

plt.show()
```

```python
df['Attrition_num'] = (df.Attrition == 'Yes').astype(int)
```

```python

df.groupby('Department', dropna=False).\
aggregate(num_rows = ('Attrition_num', 'size'),
          num_left = ('Attrition_num', 'sum'),
          attrition_rate = ('Attrition_num', 'mean')).\
reset_index()
```

```python
df.groupby('Job Role', dropna=False).\
aggregate(num_rows = ('Attrition_num', 'size'),
          num_left = ('Attrition_num', 'sum'),
          attrition_rate = ('Attrition_num', 'mean')).\
reset_index().\
sort_values('attrition_rate', ascending=False)
```

```python
sns.catplot( data = df, x='Department', hue='Attrition', kind='count', aspect=1.5 )

plt.show()
```

```python
pd.crosstab( df['Over Time'], df.Attrition )
```

```python
fig, ax = plt.subplots()

sns.heatmap( pd.crosstab( df['Over Time'], df.Attrition ), ax = ax,
             annot=True, annot_kws={'size': 16}, fmt='d',
             cmap='coolwarm', cbar=False )

plt.show()
```

```python
df.groupby('Over Time', dropna=False).\
aggregate(attrition_rate = ('Attrition_num', 'mean')).\
reset_index()
```

```python
sns.catplot( data = df, x='Attrition', y='Monthly Income', kind='box', aspect=1.3 )

plt.show()
```

```python
sns.catplot( data = df, x='Attrition', y='Monthly Income', kind='violin', aspect=1.3, inner='quartile' )

plt.show()
```

```python
df.groupby('Attrition', dropna=False).\
aggregate(income_avg = ('Monthly Income', 'mean'),
          income_median = ('Monthly Income', 'median')).\
reset_index()
```

```python
#Q5
sns.displot( data = df, x='Age', kind='hist', bins=20, aspect=1.5 )

plt.show()
```

```python
sns.displot( data = df, x='Age', kind='kde', hue='Attrition', common_norm=False, aspect=1.5 )

plt.show()
```

```python
df.groupby('Attrition', dropna=False).\
aggregate(age_avg = ('Age', 'mean'),
          tenure_avg = ('Years At Company', 'mean')).\
reset_index()
```

```python
#q6
numeric_cols = ['Attrition_num', 'Age', 'Monthly Income', 'Total Working Years', 'Job Level',
                'Job Satisfaction', 'Work Life Balance', 'Distance From Home',
                'Num Companies Worked', 'Years At Company', 'Years In Current Role',
                'Years Since Last Promotion', 'Stock Option Level']

df.loc[:, numeric_cols].corr()
```

```python
fig, ax = plt.subplots(figsize=(9, 7))

sns.heatmap( data = df.loc[:, numeric_cols].corr(),
             vmin=-1, vmax=1, center=0,
             cmap='coolwarm',
             annot=True, annot_kws={'size': 9}, fmt='.2f',
             ax=ax )

plt.show()
```
